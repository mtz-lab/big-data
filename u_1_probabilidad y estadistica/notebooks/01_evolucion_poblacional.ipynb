Actividad 3 — Estadística descriptiva

Propósito: Aplicar medidas estadísticas básicas a una variable poblacional.

```python
anio = 2020  
df_anio = df[df["Year"] == anio]
df_anio.head()
```

```python
df_extraccion_entity = df ["Entity"]
df_extraccion_entity.head(3)
```

```python
columna = "Annual change in population" 

media = df_anio[columna].mean()
mediana = df_anio[columna].median()
varianza = df_anio[columna].var()
desviacion = df_anio[columna].std()
minimo = df_anio[columna].min()
maximo = df_anio[columna].max()

print("Media:", media)
print("Mediana:", mediana)
print("Varianza:", varianza)
print("Desviación estándar:", desviacion)
print("Mínimo:", minimo)
print("Máximo:", maximo)
```

```python
entidad_min = df_anio.loc[df_anio[columna].idxmin(), "Entity"]
entidad_max = df_anio.loc[df_anio[columna].idxmax(), "Entity"]

print("Entidad con valor mínimo:", entidad_min)
print("Entidad con valor máximo:", entidad_max)
```
